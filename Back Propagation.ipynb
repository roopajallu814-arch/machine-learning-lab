```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Input columns:
# [Study Hours, Attendance]

X = np.array([
    [1, 40],
    [2, 50],
    [3, 55],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 90]
], dtype=float)

# Output:
# 0 = Fail
# 1 = Pass

y = np.array([
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
])
```

```python
# Convert values into a range between 0 and 1

X[:, 0] = X[:, 0] / 8       # Normalize study hours
X[:, 1] = X[:, 1] / 100     # Normalize attendance
```

```python
model = Sequential([
    Dense(4, activation="relu", input_shape=(2,)),
    Dense(1, activation="sigmoid")
])

# Step 5: Configure the model

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Backpropagation happens automatically during model.fit()

model.fit(
    X,
    y,
    epochs=100,
    verbose=0
)

print("Model training completed!")
```

```python
study_hours = 6
attendance = 80

new_student = np.array([
    [study_hours / 8, attendance / 100]
])

prediction = model.predict(new_student, verbose=0)

print("Prediction probability:", prediction[0][0])
```

```python
if prediction[0][0] >= 0.5:
    print("Result: Student will PASS")
else:
    print("Result: Student will FAIL")
```
