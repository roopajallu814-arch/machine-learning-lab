```python
from sklearn.tree import DecisionTreeClassifier

# Income, Age, Credit Score
X = [
    [20000, 22, 550],
    [25000, 25, 580],
    [30000, 28, 600],
    [50000, 35, 700],
    [60000, 40, 750],
    [70000, 45, 800]
]

# 0 = Poor, 1 = Good
y = [0, 0, 0, 1, 1, 1]

model = DecisionTreeClassifier(random_state=0)
model.fit(X, y)

income = int(input("Enter Income: "))
age = int(input("Enter Age: "))
credit = int(input("Enter Credit Score: "))

prediction = model.predict([[income, age, credit]])

if prediction[0] == 1:
    print("Credit Score Classification: GOOD")
else:
    print("Credit Score Classification: POOR")
```
